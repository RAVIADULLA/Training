```python
avghpbymodeldf = mtcarsdf.groupby(['model']).avg('hp')
```

```python
avghpbymodeldf.show()
```

### 4. Using OrderBy

```python
avghpbymodel_orderbydf = mtcarsdf.groupby(['model']).avg('hp').orderBy("avg(hp)")
```

```python
avghpbymodel_orderbydf.show()
```
