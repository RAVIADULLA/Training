##### https://learn.microsoft.com/en-us/azure/databricks/sql/language-manual/delta-copy-into

```python
display(dbutils.fs.ls('dbfs:/FileStore/shared_uploads/iriscloudone@outlook.com/mtcars_parquet.csv'))
```

```python
mtcarsdf = spark.read.csv('/FileStore/shared_uploads/iriscloudone@outlook.com/mtcars_parquet.csv', header=True, inferSchema=True)
```

```python
mtcarsdf.show()
mtcarsdf.printSchema()
mtcarsdf.count()
```

```python
%sql
DROP TABLE IF EXISTS mtcars_tbl;
```

```python
%sql
create table if not exists mtcars_tbl;
```

```python
display(spark.sql('select * from mtcars_tbl'))
```

```python
%sql
COPY INTO mtcars_tbl
FROM '/FileStore/shared_uploads/iriscloudone@outlook.com/mtcars_parquet.csv'
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true')
COPY_OPTIONS ('force' = 'true', 'mergeSchema' = 'true')
```

```python
display(spark.sql('select * from mtcars_tbl'))
```

```python
display(spark.sql('select model,avg(hp) from mtcars_tbl group by model'))
```
