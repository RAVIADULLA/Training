```python
%%pyspark
df = spark.read.load('abfss://processed@ssadcloudmainazdl.dfs.core.windows.net/output-from-synapse-dataflow-movies/part-00000-0bde5783-7048-455b-991a-c5188f85e66d-c000.snappy.parquet', format='parquet')
display(df.limit(10))
```

```python
df.createOrReplaceTempView('moviesView')
```

```python
moviescountbyRatingsQuery = 'SELECT count(movie),Rating FROM moviesView GROUP BY Rating'
```

```python
moviescountbyRatingsQueryResults = spark.sql(moviescountbyRatingsQuery)
```
